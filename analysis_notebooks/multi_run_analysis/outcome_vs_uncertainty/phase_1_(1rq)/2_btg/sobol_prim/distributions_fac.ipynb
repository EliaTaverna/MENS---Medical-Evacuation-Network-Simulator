# Outcome Distributions — Facility Pooling (fac=1, amb=0)

1. Waiting time for transport at CCP (T1 & T2)
2. Waiting time for care at R1 (T1 & T2)
3. Total T1 mortality distribution
4. Mortality at CCP vs mortality at R1 (compared boxplots)


```python
import sys
from pathlib import Path
_root = Path(".").resolve()
while not (_root / "paths.py").exists():
    _root = _root.parent
sys.path.insert(0, str(_root))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

RESULTS_PATH = _root / "data_cluster/saltelli_with_wrong_pooling_columns/results.csv"
df = pd.read_csv(RESULTS_PATH)
sub = df[(df["facility_eligibility"] == 1) & (df["ambulance_eligibility"] == 0)].reset_index(drop=True)
print(f"Policy: Facility Pooling (fac=1, amb=0), {len(sub)} runs")
```

## 1. Waiting Time for Transport at CCP


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, triage, color in [(axes[0], "T1", "#e74c3c"), (axes[1], "T2", "#f39c12")]:
    col = f"median_wait_ccp_transport_{triage}"
    vals = sub[col].dropna()
    ax.hist(vals, bins=30, color=color, alpha=0.7, edgecolor="white")
    ax.axvline(vals.median(), color="black", linewidth=2, linestyle="--", label=f"Median: {vals.median():.1f} min")
    ax.set_xlabel("Wait time (min)")
    ax.set_ylabel("Count")
    ax.set_title(f"{triage} — Wait for Transport at CCP")
    ax.legend()

fig.suptitle("Wait for Transport at CCP — Facility Pooling (fac=1, amb=0)", fontsize=13, y=1.02)
plt.tight_layout()
plt.show()

for t in ["T1", "T2"]:
    col = f"median_wait_ccp_transport_{t}"
    vals = sub[col].dropna()
    print(f"{t}: mean={vals.mean():.1f}, median={vals.median():.1f}, std={vals.std():.1f}, n={len(vals)}")
```

## 2. Waiting Time for Care at R1


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, triage, color in [(axes[0], "T1", "#e74c3c"), (axes[1], "T2", "#f39c12")]:
    col = f"median_wait_r1_care_{triage}"
    vals = sub[col].dropna()
    ax.hist(vals, bins=30, color=color, alpha=0.7, edgecolor="white")
    ax.axvline(vals.median(), color="black", linewidth=2, linestyle="--", label=f"Median: {vals.median():.1f} min")
    ax.set_xlabel("Wait time (min)")
    ax.set_ylabel("Count")
    ax.set_title(f"{triage} — Wait for Care at R1")
    ax.legend()

fig.suptitle("Wait for Care at R1 — Facility Pooling (fac=1, amb=0)", fontsize=13, y=1.02)
plt.tight_layout()
plt.show()

for t in ["T1", "T2"]:
    col = f"median_wait_r1_care_{t}"
    vals = sub[col].dropna()
    print(f"{t}: mean={vals.mean():.1f}, median={vals.median():.1f}, std={vals.std():.1f}, n={len(vals)}")
```

## 3. Total T1 Mortality Distribution


```python
fig, ax = plt.subplots(figsize=(10, 5))
vals = sub["death_rate_T1_BCAS"].dropna()
ax.hist(vals, bins=25, color="#e74c3c", alpha=0.7, edgecolor="white")
ax.axvline(vals.median(), color="black", linewidth=2, linestyle="--", label=f"Median: {vals.median():.3f}")
ax.set_xlabel("T1 BCAS Death Rate")
ax.set_ylabel("Count")
ax.set_title("T1 Mortality Distribution — Facility Pooling (fac=1, amb=0)")
ax.legend()
plt.tight_layout()
plt.show()

nz = (vals > 0).sum()
print(f"Mean: {vals.mean():.3f}, Median: {vals.median():.3f}, Std: {vals.std():.3f}")
print(f"Nonzero: {nz}/{len(vals)} ({nz/len(vals):.1%})")
```

## 4. Mortality by Location — CCP Transport vs R1 Care


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# T1
data_t1 = [sub["mortality_ccp_transport_T1"].dropna().values, sub["mortality_r1_care_T1"].dropna().values]
bp1 = axes[0].boxplot(data_t1, tick_labels=["CCP Transport", "R1 Care"], patch_artist=True, widths=0.5)
colors_t1 = ["#e74c3c", "#3498db"]
for patch, color in zip(bp1["boxes"], colors_t1):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)
for i, d in enumerate(data_t1):
    med = np.median(d)
    axes[0].text(i + 1, med, f"  {med:.3f}", va="center", fontsize=10, fontweight="bold")
axes[0].set_ylabel("Mortality Rate")
axes[0].set_title("T1 — Mortality by Location")

# T2
data_t2 = [sub["mortality_ccp_transport_T2"].dropna().values, sub["mortality_r1_care_T2"].dropna().values]
bp2 = axes[1].boxplot(data_t2, tick_labels=["CCP Transport", "R1 Care"], patch_artist=True, widths=0.5)
colors_t2 = ["#e74c3c", "#3498db"]
for patch, color in zip(bp2["boxes"], colors_t2):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)
for i, d in enumerate(data_t2):
    med = np.median(d)
    axes[1].text(i + 1, med, f"  {med:.3f}", va="center", fontsize=10, fontweight="bold")
axes[1].set_ylabel("Mortality Rate")
axes[1].set_title("T2 — Mortality by Location")

fig.suptitle("Where do patients die? — Facility Pooling (fac=1, amb=0)", fontsize=13, y=1.02)
plt.tight_layout()
plt.show()

for t in ["T1", "T2"]:
    for loc in ["ccp_transport", "r1_care"]:
        col = f"mortality_{loc}_{t}"
        vals = sub[col].dropna()
        nz = (vals > 0).sum()
        print(f"{t} {loc}: mean={vals.mean():.4f}, median={vals.median():.4f}, nonzero={nz}/{len(vals)} ({nz/len(vals):.1%})")
```
